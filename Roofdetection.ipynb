```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def calculate_area(contour, pixels_per_meter):
    """
    Calculate the area of a contour in square meters.
    """
    contour_area = cv2.contourArea(contour)
    area_in_meters = contour_area * (pixels_per_meter ** 2)
    return area_in_meters

def detect_rooftops(image_path, pixels_per_meter=0.25):
    """
    Detect rooftops in a satellite image and calculate their area.

    Parameters:
    - image_path (str): Path to the satellite image.
    - pixels_per_meter (float): Conversion factor from pixels to meters (default: 0.25m/pixel).

    Returns:
    - Saves an image with predicted rooftops highlighted.
    """
    # Load the input image
    image = cv2.imread(image_path)
    if image is None:
        print("Error: Image not found or invalid format.")
        return

    # Convert to grayscale for processing
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Apply edge detection to identify potential rooftops
    edges = cv2.Canny(gray, 100, 200)

    # Find contours from edges
    contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    # Create a copy of the original image to draw detected rooftops
    output_image = image.copy()

    total_rooftop_area = 0
    detected = False

    for contour in contours:
        # Approximate the contour to reduce noise
        epsilon = 0.01 * cv2.arcLength(contour, True)
        approx = cv2.approxPolyDP(contour, epsilon, True)

        # Filter based on area to detect rooftops
        contour_area = cv2.contourArea(approx)
        if contour_area > 100:  # Threshold to ignore small areas
            detected = True
            total_rooftop_area += calculate_area(approx, pixels_per_meter)
            cv2.drawContours(output_image, [approx], -1, (0, 255, 0), thickness=2)  # Draw in green

    # Save or display the output
    if detected:
        output_file = "rooftop_predictions.jpg"
        cv2.imwrite(output_file, output_image)
        print(f"Total rooftop area: {total_rooftop_area:.2f} sqm")
        print(f"Prediction saved as {output_file}.")

        # Display the image with detected rooftops
        plt.imshow(cv2.cvtColor(output_image, cv2.COLOR_BGR2RGB))
        plt.axis('off')
        plt.title(f"Detected Rooftops - Area: {total_rooftop_area:.2f} sqm")
        plt.show()
    else:
        print("No rooftops detected in the image.")
        # Save an output image indicating no rooftops were found
        output_file = "no_rooftops_detected.jpg"
        cv2.putText(output_image, "No rooftops detected", (20, 50), cv2.FONT_HERSHEY_SIMPLEX,
                    1, (0, 0, 255), 2, cv2.LINE_AA)
        cv2.imwrite(output_file, output_image)
        print(f"Output saved as {output_file}.")
        # Display the result
        plt.imshow(cv2.cvtColor(output_image, cv2.COLOR_BGR2RGB))
        plt.axis('off')
        plt.title("No Rooftops Detected")
        plt.show()

if __name__ == "__main__":
    # Example: Test with an image
    image_path = "f.png"  # Replace with your image path
    detect_rooftops(image_path)
```
